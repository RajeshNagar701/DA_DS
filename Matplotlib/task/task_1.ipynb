```python
 
import pandas as pd
import matplotlib.pyplot as plt 
import openpyxl

# Step 1: Load Excel Dataset
df = pd.read_excel("Superstore_Sales_Practice_Dataset.xlsx",
                   sheet_name="Sales_Data")

# Step 2: Convert Order Date into datetime format
df['Order Date'] = pd.to_datetime(df['Order Date'])

# Step 3: Extract Year-Month
df['Month'] = df['Order Date'].dt.to_period('M')

# Step 4: Calculate Monthly Sales
monthly_sales = df.groupby('Month')['Sales'].sum()

# Step 5: Find Highest Sales Month
highest_month = monthly_sales.idxmax()
highest_sales = monthly_sales.max()

print("Monthly Sales:")
print(monthly_sales)

print("\nHighest Sales Month:", highest_month)
print("Highest Sales Amount:", highest_sales)

# Step 6: Create Line Plot
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker='o',
    linewidth=2,
    label='Monthly Sales'
)

# Step 7: Highlight Highest Sales Month
max_index = monthly_sales.index.astype(str).tolist().index(
    str(highest_month)
)

plt.scatter(
    max_index,
    highest_sales,
    color='red',
    s=150,
    label='Highest Sales'
)

# Step 8: Add Annotation
plt.annotate(
    f'Highest Sales\n{highest_month}\n₹{highest_sales:,.2f}',
    xy=(max_index, highest_sales),
    xytext=(max_index + 2, highest_sales + 15000),
    arrowprops=dict(facecolor='red', arrowstyle='->'),
    fontsize=10,
    color='red'
)

# Step 9: Add Chart Details
plt.title('Monthly Sales Trend Analysis (2023-2024)',
          fontsize=16)

plt.xlabel('Month', fontsize=12)
plt.ylabel('Total Sales (₹)', fontsize=12)

plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

plt.tight_layout()

# Step 10: Display Chart
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel(
    "Superstore_Sales_Practice_Dataset.xlsx",
    sheet_name="Sales_Data"
)
# Group sales by category
category_sales = df.groupby('Category')['Sales'].sum()

# Sort sales in descending order
category_sales = category_sales.sort_values(ascending=False)

# Display result
print(category_sales)

# Create figure
plt.figure(figsize=(10, 6))

# Create bar chart
bars = plt.bar(
    category_sales.index,
    category_sales.values,
    color=['royalblue', 'orange', 'green'],
    width=0.5,
    edgecolor='black'
)

# Display sales values on top of each bar
for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f'₹{height:,.0f}',
        ha='center',
        va='bottom',
        fontsize=11,
        fontweight='bold'
    )

# Customize chart
plt.title(
    'Category-wise Sales Comparison',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Total Sales (₹)', fontsize=12)

plt.xticks(fontsize=11)
plt.yticks(fontsize=10)

plt.grid(axis='y', linestyle='--', alpha=0.4)

# Remove unnecessary top and right borders
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()

# Display chart
plt.show()
```

```python
#task 3

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel(
    "Superstore_Sales_Practice_Dataset.xlsx",
    sheet_name="Sales_Data"
)

df.head()

# Calculate mean sales
mean_sales = df['Sales'].mean()

# Calculate median sales
median_sales = df['Sales'].median()

print("Mean Sales:", mean_sales)
print("Median Sales:", median_sales)

# Calculate mean and median
mean_sales = df['Sales'].mean()
median_sales = df['Sales'].median()

# Create histogram
plt.figure(figsize=(12, 6))

plt.hist(
    df['Sales'],
    bins=20,
    color='skyblue',
    edgecolor='black',
    alpha=0.8
)

# Add Mean reference line
plt.axvline(
    mean_sales,
    color='red',
    linestyle='--',
    linewidth=2,
    label=f'Mean = ₹{mean_sales:,.2f}'
)

# Add Median reference line
plt.axvline(
    median_sales,
    color='green',
    linestyle='-',
    linewidth=2,
    label=f'Median = ₹{median_sales:,.2f}'
)

# Add chart title and labels
plt.title(
    'Sales Distribution Analysis',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Sales Amount (₹)', fontsize=12)
plt.ylabel('Frequency (Number of Orders)', fontsize=12)

# Add legend
plt.legend(fontsize=11)

# Add grid
plt.grid(axis='y', linestyle='--', alpha=0.4)

# Display chart
plt.tight_layout()
plt.show()
```

```python
#task 4

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel(
    "Superstore_Sales_Practice_Dataset.xlsx",
    sheet_name="Sales_Data"
)

# Display first 5 records
df.head()



regional_sales = df.groupby('Region')['Sales'].sum()

# Step 2: Calculate Regional Profit
regional_profit = df.groupby('Region')['Profit'].sum()

# Step 3: Calculate Monthly Sales
df['Order Date'] = pd.to_datetime(df['Order Date'])

df['Month'] = df['Order Date'].dt.to_period('M')

monthly_sales = df.groupby('Month')['Sales'].sum()

# Step 4: Create 2x2 Dashboard
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Main Dashboard Title
fig.suptitle(
    'REGIONAL SALES PERFORMANCE DASHBOARD',
    fontsize=20,
    fontweight='bold'
)

# ==========================================
# CHART 1: REGIONAL SALES
# ==========================================

axes[0, 0].bar(
    regional_sales.index,
    regional_sales.values,
    color='royalblue',
    edgecolor='black'
)

axes[0, 0].set_title(
    'Regional Sales Comparison',
    fontsize=14,
    fontweight='bold'
)

axes[0, 0].set_xlabel('Region')
axes[0, 0].set_ylabel('Total Sales (₹)')

# Display values on bars
for i, value in enumerate(regional_sales.values):
    axes[0, 0].text(
        i,
        value,
        f'₹{value:,.0f}',
        ha='center',
        va='bottom',
        fontsize=9
    )

axes[0, 0].grid(axis='y', linestyle='--', alpha=0.4)


# ==========================================
# CHART 2: REGIONAL PROFIT
# ==========================================

axes[0, 1].bar(
    regional_profit.index,
    regional_profit.values,
    color='seagreen',
    edgecolor='black'
)

axes[0, 1].set_title(
    'Regional Profit Comparison',
    fontsize=14,
    fontweight='bold'
)

axes[0, 1].set_xlabel('Region')
axes[0, 1].set_ylabel('Total Profit (₹)')

# Display values on bars
for i, value in enumerate(regional_profit.values):
    axes[0, 1].text(
        i,
        value - 1500,
        f'₹{value:,.0f}',
        ha='center',
        va='top',
        fontsize=9
    )

axes[0, 1].axhline(0, color='black', linewidth=0.8)
axes[0, 1].grid(axis='y', linestyle='--', alpha=0.4)


# ==========================================
# CHART 3: MONTHLY SALES TREND
# ==========================================

axes[1, 0].plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    color='orange',
    marker='o',
    linewidth=2,
    markersize=5
)

axes[1, 0].set_title(
    'Monthly Sales Trends (2023-2024)',
    fontsize=14,
    fontweight='bold'
)

axes[1, 0].set_xlabel('Month')
axes[1, 0].set_ylabel('Monthly Sales (₹)')

axes[1, 0].tick_params(axis='x', rotation=60)
axes[1, 0].grid(True, linestyle='--', alpha=0.4)


# ==========================================
# CHART 4: SALES VS PROFIT
# ==========================================

for region in df['Region'].unique():

    region_data = df[df['Region'] == region]

    axes[1, 1].scatter(
        region_data['Sales'],
        region_data['Profit'],
        label=region,
        alpha=0.6,
        s=35
    )

axes[1, 1].set_title(
    'Sales vs Profit Analysis',
    fontsize=14,
    fontweight='bold'
)

axes[1, 1].set_xlabel('Sales (₹)')
axes[1, 1].set_ylabel('Profit (₹)')

axes[1, 1].axhline(
    0,
    color='red',
    linestyle='--',
    linewidth=1
)

axes[1, 1].legend(title='Region')
axes[1, 1].grid(True, linestyle='--', alpha=0.4)


# ==========================================
# FINAL DASHBOARD SETTINGS
# ==========================================

plt.tight_layout(rect=[0, 0, 1, 0.95])

# Display Dashboard
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Excel dataset
df = pd.read_excel(
    "Superstore_Sales_Practice_Dataset.xlsx",
    sheet_name="Sales_Data"
)

# Display first 5 records
df.head()

# Create figure
plt.figure(figsize=(12, 7))

# Get unique product categories
categories = df['Category'].unique()

# Plot each category with a different color
for category in categories:

    category_data = df[df['Category'] == category]

    plt.scatter(
        category_data['Quantity'],
        category_data['Sales'],
        label=category,
        alpha=0.6,
        s=50
    )

# Add chart title and labels
plt.title('Quantity vs Sales Analysis', fontsize=16, fontweight='bold')
plt.xlabel('Quantity', fontsize=12)
plt.ylabel('Sales Amount', fontsize=12)

# Display legend
plt.legend(title='Product Category')

# Add grid
plt.grid(alpha=0.3)

# Display chart
plt.show()
```
